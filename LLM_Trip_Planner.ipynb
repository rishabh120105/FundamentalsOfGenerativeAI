```python
import os
import requests
from groq import Groq

# Initialize Groq client
client = Groq(api_key=os.environ["GROQ_API_KEY"])


# -------------------------
# WEATHER TOOL
# -------------------------
def get_weather(city):
    api_key = os.environ["OPENWEATHER_API_KEY"]
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"

    try:
        data = requests.get(url).json()
        temp = data["main"]["temp"]
        desc = data["weather"][0]["description"]
        return f"{city}: {temp}°C, {desc}"
    except:
        return "Weather unavailable"


# -------------------------
# FLIGHTS (Demo)
# -------------------------
def flights(city):
    return f"""
Flights to {city}:
- Air India (~₹35k)
- Emirates (~₹48k)
- Singapore Airlines (~₹42k)
"""


# -------------------------
# HOTELS (Demo)
# -------------------------
def hotels(city):
    return f"""
Hotels in {city}:
- Budget Inn (~₹3000/night)
- Comfort Stay (~₹7000/night)
- Grand Palace (~₹15000/night)
"""


# -------------------------
# MAIN TRIP PLANNER
# -------------------------
def trip_planner(query):

    city = query.split("to")[-1].strip().split(" ")[0]

    weather = get_weather(city)
    flight_info = flights(city)
    hotel_info = hotels(city)

    prompt = f"""
You are an AI travel planner.

Generate:

1. One paragraph about cultural & historical significance
2. Current weather:
{weather}
3. Suggested travel dates
4. Flight options:
{flight_info}
5. Hotel options:
{hotel_info}
6. Day-wise itinerary

User request:
{query}
"""

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content


# -------------------------
# RUN TEST
# -------------------------
query = "Plan a 3-day trip to Tokyo in May"
print(trip_planner(query))
```
